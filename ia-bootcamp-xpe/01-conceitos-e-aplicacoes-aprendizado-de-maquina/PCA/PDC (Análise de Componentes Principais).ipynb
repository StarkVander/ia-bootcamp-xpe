# Análise de Componentes Principais (PCA)

## Introdução

Em muitos problemas do mundo real, os conjuntos de dados costumam conter várias variáveis ou características. No entanto, nem todas essas variáveis são igualmente relevantes para a análise. Além disso, é difícil visualizar e interpretar dados que possuem mais de três dimensões.

O PCA permite reduzir a dimensionalidade dos dados, mantendo a maior parte da informação relevante contida nas variáveis originais. Ele faz isso transformando as variáveis originais em um novo conjunto de variáveis, **chamadas de componentes principais**. Esses componentes são combinações lineares das variáveis originais e são ordenados de acordo com a quantidade de variação explicada que eles capturam.

![Ilustração PCA](./img/pca.jpg)

A ideia principal do PCA é que as informações mais importantes contidas nos dados são aquelas relacionadas às direções de maior variação. **Os primeiros componentes principais capturam a maior parte da variância nos dados, enquanto os componentes subsequentes capturam cada vez menos variancia**. Dessa forma, podemos escolher reter apenas os primeiros componentes principais mais significativos e descartar os menos relevantes, reduzindo assim a dimensionalidade dos dados.

### Importando as bibliotecas
***

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Carregar o conjunto de dados iris
iris = load_iris()
print(iris.DESCR)
```

### Estruturando base de dados
***

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['class'] = iris.target
```

```python
df.head()
```

### Mapeamento variável target
***

```python
class_map = {
    0 : 'Iris-Setosa',
    1 : 'Iris-Versicolour',
    2 : 'Iris-Virginica',
}

df['class'] = df['class'].map(class_map)
```
